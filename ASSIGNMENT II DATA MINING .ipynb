# List all products

```python
transactions = []

for i in range(len(data)):
    
    transaction = []
    
    for item in data.iloc[i]:
        
        if pd.notna(item):
            transaction.append(str(item).strip())
    
    transactions.append(transaction)

print(transactions[:5])
```

# ==================================
# Question 2: Frequent Itemsets
# ==================================

# ------------------------------------------------
# Top 10 Most Frequently Purchased Items
# ------------------------------------------------

# Count how many transactions contain each item and Remove duplicate items in the same transaction

```python
item_count = {}

for transaction in transactions:
    unique_items = set(transaction)

    for item in unique_items:

        if item in item_count:
            item_count[item] = item_count[item] + 1
        else:
            item_count[item] = 1
```

# Sort items from highest to lowest

```python
sorted_items = sorted(
    item_count.items(),
    key=lambda x: x[1],
    reverse=True
)
```

# Take the top 10 items

```python
top10 = sorted_items[:10]
top10_table = pd.DataFrame(
    top10,
    columns=["Item", "Number of Transactions"]
)
print("Top 10 Most Frequently Purchased Items")
display(top10_table)
```

```python
# -------------------------------
# STEP 4: Association Rules
# -------------------------------

print("=" * 60)
print("QUESTION 3: ASSOCIATION RULES")
print("=" * 60)

# Generate association rules
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.2
)

# Sort by highest lift, then highest confidence
rules = rules.sort_values(
    by=["lift", "confidence"],
    ascending=False
)

# Select the important columns
selected_rules = rules[
    [
        "antecedents",
        "consequents",
        "support",
        "confidence",
        "lift"
    ]
].copy()

# Select the top 10 rules
top10_rules = selected_rules.head(10).copy()

# Round the numerical values
top10_rules["support"] = top10_rules["support"].round(4)
top10_rules["confidence"] = top10_rules["confidence"].round(4)
top10_rules["lift"] = top10_rules["lift"].round(4)

# Add ranking
top10_rules.insert(0, "Rank", range(1, 11))

# Display the table
display(top10_rules)
```

```python
print("=" * 60)
print("QUESTION 4: UNDERSTANDING THE MEASURES")
print("=" * 60)
```

# Rule 1: {beef} → {whole milk}
# Support = 0.02
# Confidence = 0.41
# Lift = 1.59

# a. What does the support mean?
## A support of 0.02 (2%) means that 2% of all transactions in the dataset contained both beef and whole milk together.

# b. What does the confidence mean?
## A confidence of 0.41 (41%) means that 41% of customers who purchased beef also purchased whole milk.

# c. What does the lift mean?
## A lift of 1.59 means that customers who buy beef are 1.59 times more likely to buy whole milk than a customer selected at random. 
## Since the lift is greater than 1, there is a positive association between beef and whole milk.

# Rule 2: {butter} → {other vegetables}
# Support = 0.02
# Confidence = 0.36
# Lift = 1.87

# a. What does the support mean?
## A support of 0.02 (2%) indicates that 2% of all transactions included both butter and other vegetables.


# b. What does the confidence mean?
## A confidence of 0.36 (36%) means that 36% of customers who bought butter also bought other vegetables.

# c. What does the lift mean?
## A lift of 1.87 shows that customers who buy butter are 1.87 times more likely to buy other vegetables than expected by chance. 
## This indicates a strong positive relationship between these products.

# Rule 3: {butter} → {whole milk}
# Support = 0.03
# Confidence = 0.50
# Lift = 1.95

# a. What does the support mean?
## A support of 0.03 (3%) means that 3% of all transactions contained both butter and whole milk.

# b. What does the confidence mean?
## A confidence of 0.50 (50%) means that 50% of customers who purchased butter also purchased whole milk.

# c. What does the lift mean?
## A lift of 1.95 means that customers who buy butter are 1.95 times more likely to buy whole milk than a randomly selected customer. 
## Because the lift is close to 2, this is the strongest association among the three selected rules.

```python
# -------------------------------
# STEP 5: Highest Metrics Rules
# -------------------------------

print("=" * 60)
print("QUESTION 5: HIGHEST ASSOCIATION RULE METRICS")
print("=" * 60)

# Find the rule with the highest confidence
highest_confidence_rule = rules.loc[
    rules["confidence"].idxmax()
]

# Find the rule with the highest lift
highest_lift_rule = rules.loc[
    rules["lift"].idxmax()
]

# Find the rule with the highest support
highest_support_rule = rules.loc[
    rules["support"].idxmax()
]

# Create a table
question5 = pd.DataFrame({
    "Metric": [
        "Highest Confidence",
        "Highest Lift",
        "Highest Support"
    ],
    
    "Antecedents": [
        highest_confidence_rule["antecedents"],
        highest_lift_rule["antecedents"],
        highest_support_rule["antecedents"]
    ],
    
    "Consequents": [
        highest_confidence_rule["consequents"],
        highest_lift_rule["consequents"],
        highest_support_rule["consequents"]
    ],
    
    "Support": [
        highest_confidence_rule["support"],
        highest_lift_rule["support"],
        highest_support_rule["support"]
    ],
    
    "Confidence": [
        highest_confidence_rule["confidence"],
        highest_lift_rule["confidence"],
        highest_support_rule["confidence"]
    ],
    
    "Lift": [
        highest_confidence_rule["lift"],
        highest_lift_rule["lift"],
        highest_support_rule["lift"]
    ]
})

# Round the numerical values
question5["Support"] = question5["Support"].round(4)
question5["Confidence"] = question5["Confidence"].round(4)
question5["Lift"] = question5["Lift"].round(4)

# Display the results
display(question5)
```

```python
# STEP 5: Highest Metrics Rules

print("=" * 60)
print("QUESTION 5: HIGHEST METRICS")
print("=" * 60)

# -----------------------------------
# 1. Highest Confidence
# -----------------------------------

rule = rules.loc[rules["confidence"].idxmax()]

print("\nHighest Confidence Rule")
print("-----------------------")
print("From:", rule["antecedents"])
print("To:", rule["consequents"])
print("Support:", round(rule["support"], 4))
print("Confidence:", round(rule["confidence"], 4))
print("Lift:", round(rule["lift"], 4))


# -----------------------------------
# 2. Highest Lift
# -----------------------------------

rule = rules.loc[rules["lift"].idxmax()]

print("\nHighest Lift Rule")
print("-----------------")
print("From:", rule["antecedents"])
print("To:", rule["consequents"])
print("Support:", round(rule["support"], 4))
print("Confidence:", round(rule["confidence"], 4))
print("Lift:", round(rule["lift"], 4))


# -----------------------------------
# 3. Highest Support
# -----------------------------------

rule = rules.loc[rules["support"].idxmax()]

print("\nHighest Support Rule")
print("--------------------")
print("From:", rule["antecedents"])
print("To:", rule["consequents"])
print("Support:", round(rule["support"], 4))
print("Confidence:", round(rule["confidence"], 4))
print("Lift:", round(rule["lift"], 4))
```
